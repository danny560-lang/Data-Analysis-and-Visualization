## Inspect and Select Numeric Columns

<div class="alert alert-block alert-info">
    <span style="color:black"><b><i>Hints:</i></b> Use <b><mark>select_dtypes()</mark></b> to filter columns with numeric data types (float64 and int64). Removes columns like <b><mark>UNITID</mark></b> and <b><mark>OPEID6</mark></b>, which are not useful for clustering. Use <b><mark>dropna()</mark></b> to remove rows with missing values to clean data for clustering, see also module three sample for additional cleaning examples. Use error handling <b><mark>try blocks</mark></b> to ensures the program stops if the column selection fails.</span>
</div>

```python
print("Inspecting dataset to select numeric columns for clustering...")
# To Do: In a try/catch block, exclude identifiers 'UNITID' and 'OPEID6', 
#        then print the numeric columns selected for clustering and also print the data for clustering in rows & columns
try:
    #select numeric data with types (float64 and int64)
    numeric_data = data.select-dtypes(includes = ['float64', 'int64'])
    
    #Remove columns that are not useful for clustering
    numeric_data = numeric_data.drop(columns = ['UNITID', 'OPEID6'], errors = 'ignore')
    
    #Remove rows with missing values to clean data
    numeric_data = numeric_data.dropna()
    
    print(f"umeric columns selected for clustering: {numeric_data.columns.tolist()}")
    print(f"Data shape for clustering: {numeric_data.shape[0]} rows x {numeric_data.shape[1]} columns")
except Exception as e:
    print(f"Error selecting numeric columns: {e}")
    raise
```

## Normalize the Data

<div class="alert alert-block alert-info">
    <span style="color:black"><b><i>Hints:</i></b> Use <b><mark>MinMaxScaler()</mark></b> to scale data values between 0 and 1, ensuring all features are on the same scale. Use <b><mark>fit_transform()</mark></b> to fit the scaler to the data and transform it.</span>
</div>

```python
print("Normalizing the data using MinMaxScaler...")
# To Do: in a try/catch block, normalize the data using MinMaxScaler
try:
    scaler = MinMaxScaler()
    scaled_data = scaler.fit_transform(numeric_data)
    print(f"Scaled data shape: {scaled_data.shape}")
    print("Data successfully standardized (mean≈0, std≈1 per feature)")
except Exception as e:
    print(f"Error during scaling: {e}")
    raise
```

## Perform Hierarchical Clustering

<div class="alert alert-block alert-info">
    <span style="color:black"><b><i>Hints:</i></b> Use <b><mark>linkage()</mark></b> to compute hierarchical clusters using <b><mark>Ward's</mark></b> method, which minimizes variance between clusters. Limit clustering to the first 1000 rows for performance. Use <b><mark>fcluster()</mark></b> to assign cluster labels to data points based on the number of clusters (max_clusters).</span>
</div>

```python
print("Performing hierarchical clustering using Ward's method...")
# To Do: in a try/catch block use the Ward's method to perform hierarchical clustering, limit max clusters to 4.
try:
    # Limit to first 1000 rows for performance
    sample_data = scaled_data[:1000]

    # Compute hierarchical clusters using Ward's method
    linked = linkage(sample_data, method='ward')

    # Assign cluster labels based on max number of clusters
    max_clusters = 4
    cluster_labels = fcluster(linked, max_clusters, criterion='maxclust')

    print(f"Clustering complete. {max_clusters} clusters assigned to {len(cluster_labels)} records.")
    print(f"Cluster label counts: { {i: list(cluster_labels).count(i) for i in range(1, max_clusters + 1)} }")
except Exception as e:
    print(f"Error during hierarchical clustering: {e}")
    raise
```

### Dendrogram

<div class="alert alert-block alert-info">
    <span style="color:black"><b><i>Hints:</i></b> A dendrogram displays the hierarchical structure of clusters. Save the image as <b><mark>dendrogram.png</mark></b>. Use <b><mark>display()</mark></b> to render the image for html file output. Do not use an <mark>html image</mark> tag in the markdown, or the image will not print in the html download.</span>
</div>

```python
print("Creating dendrogram to visualize hierarchical clustering...")
# To Do: Create a dendrogram to visualize hierarchical clustering, and save the dendrogram into a png file under the 'output' folder.

# Sample 80 rows so the plot remains readable
sample_idx = np.random.choice(len(scaled_data), 80, replace=False)
sample_scaled = scaled_data[sample_idx]
linked = linkage(sample_scaled, method='ward')

fig, ax = plt.subplots(figsize=(14, 6))
dendrogram(
    linked, ax=ax,
    truncate_mode='lastp', p=20,
    color_threshold=0.7 * max(linked[:, 2]),
    leaf_rotation=45, leaf_font_size=9
)
ax.set_title(
    'Hierarchical Clustering Dendrogram\n(Ward Linkage – College Scorecard Features)',
    fontsize=14, fontweight='bold', pad=15
)
ax.set_xlabel('Sample Clusters (Merged Leaves)', fontsize=11)
ax.set_ylabel('Distance (Ward)', fontsize=11)
ax.axhline(
    y=0.7 * max(linked[:, 2]),
    color='red', linestyle='--', linewidth=1.2, label='Cut-off threshold'
)
ax.legend(fontsize=10)
plt.tight_layout()
plt.savefig('output/dendrogram.png', dpi=150, bbox_inches='tight')
plt.close()
print("Dendrogram saved to output/dendrogram.png")
```

### PCA Scatter Plot

<div class="alert alert-block alert-info">
    <span style="color:black"><b><i>Hints:</i></b> A PCA scatter plot projects high-dimensional data into 2D using PCA for visualization. Save the image as <b><mark>pca_scatter_plot.png</mark></b>. Use <b><mark>display()</mark></b> to render the image for html file output. Do not use an <mark>html image</mark> tag in the markdown, or the image will not print in the html download.</span>
</div>

```python
print("Creating scatter plot (PCA Reduced) to visualize clusters in 2D...")
# To Do: Create a PCA Scatter Plot to visualize clusters in 2D, and save the PCA Scatter Plot into a png file under the 'output' folder.
pca = PCA(n_components=2, random_state=42)
pca_result = pca.fit_transform(scaled_data)
pca_df = pd.DataFrame(pca_result, columns=['PC1', 'PC2'])
pca_df['Cluster'] = cluster_labels

var1 = pca.explained_variance_ratio_[0] * 100
var2 = pca.explained_variance_ratio_[1] * 100
palette = {0: '#2196F3', 1: '#FF9800', 2: '#4CAF50', 3: '#E91E63'}

fig, ax = plt.subplots(figsize=(10, 7))
for c, grp in pca_df.groupby('Cluster'):
    ax.scatter(
        grp['PC1'], grp['PC2'],
        c=palette[c], label=f'Cluster {c}',
        alpha=0.7, s=55, edgecolors='white', linewidths=0.5
    )
ax.set_title(
    'PCA Scatterplot – Cluster Separation\n(College Scorecard Field of Study)',
    fontsize=14, fontweight='bold', pad=15
)
ax.set_xlabel(f'Principal Component 1 ({var1:.1f}% variance)', fontsize=11)
ax.set_ylabel(f'Principal Component 2 ({var2:.1f}% variance)', fontsize=11)
ax.legend(title='Cluster', title_fontsize=11, fontsize=10, framealpha=0.9)
ax.grid(True, linestyle='--', alpha=0.4)
plt.tight_layout()
plt.savefig('output/pca_scatterplot.png', dpi=150, bbox_inches='tight')
plt.close()
print("PCA scatterplot saved to output/pca_scatterplot.png")

```

### Cluster Size Distribution

<div class="alert alert-block alert-info">

```python
print("Creating cluster size distribution chart to analyze cluster populations...")
# To Do: Create a cluster size distribution chart, and save it into a png file under the 'output' folder.
cluster_counts = numeric_data['Cluster'].value_counts().sort_index()
colors = ['#2196F3', '#FF9800', '#4CAF50', '#E91E63']

fig, ax = plt.subplots(figsize=(8, 6))
bars = ax.bar(
    [f'Cluster {i}' for i in cluster_counts.index],
    cluster_counts.values,
    color=colors, edgecolor='white', linewidth=1.5, alpha=0.9
)
for bar, val in zip(bars, cluster_counts.values):
    ax.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height() + 5, str(val),
        ha='center', va='bottom', fontsize=12, fontweight='bold'
    )
ax.set_title(
    'Cluster Size Distribution\n(KMeans, k=4 – College Scorecard Data)',
    fontsize=14, fontweight='bold', pad=15
)
ax.set_xlabel('Cluster', fontsize=12)
ax.set_ylabel('Number of Records', fontsize=12)
ax.set_ylim(0, max(cluster_counts.values) * 1.15)
ax.grid(axis='y', linestyle='--', alpha=0.4)
ax.set_facecolor('#FAFAFA')
plt.tight_layout()
plt.savefig('output/cluster_size_distribution.png', dpi=150, bbox_inches='tight')
plt.close()
print("Cluster size distribution chart saved.")
```

### Trend Chart

<div class="alert alert-block alert-info">
    <span style="color:black"><b><i>Hints:</i></b> A trend chart plots trends in a numeric feature across clusters. Save the image as <b><mark>trend_chart.png</mark></b>. Use <b><mark>display()</mark></b> to render the image for html file output. Do not use an <mark>html image</mark> tag in the markdown, or the image will not print in the html download.</span>
</div>

```python
print("Creating trend chart to analyze a numeric feature across clusters...")
# To Do: Create a trend chart to analyze a numeric feature across clusters, and save it into a png file under the 'output' folder.
cluster_earn = numeric_data.groupby('Cluster')[
    ['EARN_MDN_HI_1YR', 'EARN_MDN_HI_2YR']
].mean()

x = np.arange(len(cluster_earn))
width = 0.35

fig, ax = plt.subplots(figsize=(10, 6))
bars1 = ax.bar(
    x - width / 2, cluster_earn['EARN_MDN_HI_1YR'], width,
    label='Median Earnings Year 1', color='#42A5F5', alpha=0.9, edgecolor='white'
)
bars2 = ax.bar(
    x + width / 2, cluster_earn['EARN_MDN_HI_2YR'], width,
    label='Median Earnings Year 2', color='#FF7043', alpha=0.9, edgecolor='white'
)
for bar in bars1:
    ax.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height() + 400, f'${bar.get_height():,.0f}',
        ha='center', va='bottom', fontsize=8
    )
for bar in bars2:
    ax.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height() + 400, f'${bar.get_height():,.0f}',
        ha='center', va='bottom', fontsize=8
    )

# Overlay trend lines
ax.plot(x - width / 2, cluster_earn['EARN_MDN_HI_1YR'].values,
        'o--', color='#1565C0', linewidth=1.5, markersize=6, alpha=0.8)
ax.plot(x + width / 2, cluster_earn['EARN_MDN_HI_2YR'].values,
        's--', color='#BF360C', linewidth=1.5, markersize=6, alpha=0.8)

ax.set_xticks(x)
ax.set_xticklabels([f'Cluster {i}' for i in cluster_earn.index], fontsize=11)
ax.set_title(
    'Earnings Trend by Cluster: Year 1 vs Year 2\n'
    '(Median Post-Graduation Earnings – College Scorecard)',
    fontsize=14, fontweight='bold', pad=15
)
ax.set_xlabel('Cluster', fontsize=12)
ax.set_ylabel('Median Earnings (USD)', fontsize=12)
ax.legend(fontsize=10, framealpha=0.9)
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: f'${v:,.0f}'))
ax.grid(axis='y', linestyle='--', alpha=0.4)
ax.set_facecolor('#FAFAFA')
plt.tight_layout()
plt.savefig('output/trend_chart.png', dpi=150, bbox_inches='tight')
plt.close()
print("Trend chart saved.")

```
